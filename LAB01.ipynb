Ejercicio 2. Desarrollar una funci´on triangSup(A) que devuelva la matriz U correspondiente a la matriz Triangular Superior de A sin su diagonal.

```python
def triangSup(A):
    n = len(A)
    
    # Recorrer cada columna (excepto la última)
    for i in range(n - 1):
        # Evitar división por cero buscando un pivote diferente de cero
        if A[i][i] == 0:
            for k in range(i + 1, n):
                if A[k][i] != 0:
                    A[i], A[k] = A[k], A[i]  # Intercambiar filas
                    break
                    
        # Hacer ceros debajo del pivote actual
        for j in range(i + 1, n):
            factor = A[j][i] / A[i][i]
            for k in range(i, n):
                A[j][k] = A[j][k] - factor * A[i][k]
                
    return A

matriz_original = np.array([
    [2.0,  1.0, -1.0],
    [-3.0, -1.0,  2.0],
    [-2.0,  1.0,  2.0]
])

triangSup(matriz_original)
```

Ejercicio 3. Desarrollar una funci´on triangInf(A) que devuelva la matriz L correspondiente a la matriz Triangular Inferior de A sin su diagonal.

```python
def triangInf(A):
    return triangSup(A)[::-1]

matriz_original = np.array([
    [2.0,  1.0, -1.0],
    [-3.0, -1.0,  2.0],
    [-2.0,  1.0,  2.0]
])

triangInf(matriz_original)
```

Ejercicio 4. Desarrollar una funci´on diagonal(A) que devuelva la matriz D
correspondiente a la matriz diagonal de A.

```python
def diagonal(A):
    triang = triangSup(A)
    n = len(A)

    for i in range(n-1):
        for j in range (i+1, n):
            triang[i][j] = 0

    return triang

diagonal(matriz_original)
```

Ejercicio 5. Desarrollar una funci´on traza(A) que calcule la traza de una
matriz cualquiera A.

```python
def traza(A):
    ans = 0
    n = len(A)

    for k in range(n):
        ans += sum(A[k])
    
    return ans

traza(matriz_original)
```

Ejercicio 6. Desarrollar una funci´on traspuesta(A) que devuelva la matriz
traspuesta de A.

```python
def traspuesta(A):
    A_array = np.array(A)
    A_copy = A_array.copy()

    n = len(A_copy)
    for i in range(n):
        for j in range(i + 1, n):
            A_copy[i, j], A_copy[j, i] = A_copy[j, i], A_copy[i, j]

    return A_copy

matriz_original = np.array([
    [2.0,  1.0, -1.0],
    [-3.0, -1.0,  2.0],
    [-2.0,  1.0,  2.0]
])

print(matriz_original)
print("Traspuesta:")
print(traspuesta(matriz_original))
```

Ejercicio 7. Desarrollar una funci´on esSimetrica(A) que devuelve True si la
matriz A es sim´etrica y False en caso contrario.

```python
def esSimetrica(A):
    if not esCuadrada(A):
        return False
    
    n = len(A)

    for i in range(n):
        for j in range (n):
            if A[i][j] != A[j][i]:
                return False
    
    d = A[0][0]
    for i in range(n):
        if A[i][i] != d:
            return False

    return True

esSimetrica(matriz_original)
```
